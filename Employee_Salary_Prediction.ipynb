# **PART1-DATA PREPROCESSING**

```python
#Load the Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sn
import sklearn
```

```python
data=pd.read_csv("../dataset/employee_performance.csv")
```

```python
data.head()
```

```python
#Handle Missing Values
data.isnull().sum()
```

```python
data['Experience_Years'] = data['Experience_Years'].fillna(
    data['Experience_Years'].median())
data['Performance_Score'] = data['Performance_Score'].fillna(
    data['Performance_Score'].median()
)
```

```python
# Remove Duplicates
data.isnull().sum()
```

```python
#Remove Duplicates
data.duplicated().sum()
```

```python
data=data.drop_duplicates()
```

```python
data.duplicated().sum()
```

```python
# Encode Categorical Features
data.select_dtypes(include='object').columns
```

```python
data=pd.get_dummies(data,
                    columns=['Gender','Department'],
                    drop_first=True,
                    dtype=int)
```

```python
data.head()
```

```python
# Scale Numerical Features
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
data[['Age','Experience_Years','Performance_Score']] = scaler.fit_transform(
    data[['Age','Experience_Years','Performance_Score']])
```

```python
data[['Age','Experience_Years','Performance_Score']].head()
```

```python
#Split the dataset into training and testing sets
from sklearn.model_selection import train_test_split
X= data.drop(['Employee_ID','Salary'],axis=1)
Y=data['Salary']
X_train,X_test,Y_train,Y_test=train_test_split(
    X,Y,test_size=0.2,
random_state=42)
```

```python
print(X_train.shape)
```

```python
print(X_test.shape)
```

```python
print(Y_train.shape)
```

```python
print(Y_test.shape)
```

# **PART 2-MODEL DEVELOPMENT**

```python
# Linear Regression
from sklearn.linear_model import LinearRegression
model_1r=LinearRegression()
model_1r.fit(X_train,Y_train)
```

```python
# Decision Tree Regressor
from sklearn.tree import DecisionTreeRegressor
model_dt=DecisionTreeRegressor(random_state=42)
model_dt.fit(X_train,Y_train)
```

```python
# Random Forest Regressor
from sklearn.ensemble import RandomForestRegressor
model_rf=RandomForestRegressor(random_state=42)
model_rf.fit(X_train,Y_train)
```

# **PART 3-MODEL EVALUATION**

```python
Y_pred_1r=model_1r.predict(X_test)
Y_pred_dt=model_dt.predict(X_test)
Y_pred_rf=model_rf.predict(X_test)
```

```python
# MAE
from sklearn.metrics import mean_absolute_error
mae_1r=mean_absolute_error(Y_test,Y_pred_1r)
mae_dt=mean_absolute_error(Y_test,Y_pred_dt)
mae_rf=mean_absolute_error(Y_test,Y_pred_rf)
print("Linear Regression MAE:",mae_1r)
print("Decision Tree MAE:",mae_dt)
print("Random Forest MAE:",mae_rf)
```

```python
#MSE
from sklearn.metrics import mean_squared_error
mse_1r=mean_squared_error(Y_test,Y_pred_1r)
mse_dt=mean_squared_error(Y_test,Y_pred_dt)
mse_rf=mean_squared_error(Y_test,Y_pred_rf)
print("Linear Regression MSE:",mse_1r)
print("Decision Tree MSE:",mse_dt)
print("Random Forest MSE:",mse_rf)
```

```python
#RMSE
import numpy as np
rmse_1r=np.sqrt(mse_1r)
rmse_dt=np.sqrt(mse_dt)
rmse_rf=np.sqrt(mse_rf)
print("Linear Regression RMSE:",rmse_1r)
print("Decision Tree RMSE:",rmse_dt)
print("Random Forest RMSE:",rmse_rf)
```

```python
# R^2 Score
from sklearn.metrics import r2_score
r2_1r=r2_score(Y_test,Y_pred_1r)
r2_dt=r2_score(Y_test,Y_pred_dt)
r2_rf=r2_score(Y_test,Y_pred_rf)
print("Linear Regression R2 Score:",r2_1r)
print("Decision Tree R2 Score:",r2_dt)
print("Random Forest R2 Score:",r2_rf)
```

# **PART4-MODLE COMPARISON**

```python
# Training Time + Comparison Table
import time
start = time.time()
model_1r.fit(X_train, Y_train)
time_1r = time.time() - start
start = time.time()
model_dt.fit(X_train, Y_train)
time_dt = time.time() - start
start = time.time()
model_rf.fit(X_train, Y_train)
time_rf = time.time() - start
print("Linear Regression Training Time:", time_1r)
print("Decision Tree Training Time:", time_dt)
print("Random Forest Training Time:", time_rf)
```

```python
#Model Comparison Table
comparison = {
    "Model Name": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest" ],
    "MAE": [
        mae_1r,
        mae_dt,
        mae_rf ],
    "MSE": [
        mse_1r,
        mse_dt,
        mse_rf],
    "RMSE": [
        rmse_1r,
        rmse_dt,
        rmse_rf ],
    "R2 Score": [
        r2_1r,
        r2_dt,
        r2_rf],
    "Training Time": [
        time_1r,
        time_dt,
        time_rf]
}

comparison
```

```python
#Advantages & Limitations
comparison["Advantages"] = [
    "Simple and fast",
    "Easy to understand",
    "Good prediction performance"]
comparison["Limitations"] = [
    "Assumes linear relationship",
    "Can overfit",
    "More complex and slower"]

comparison
```

```python
#Best Model Finding
print("Linear Regression R2:", r2_1r)
print("Decision Tree R2:", r2_dt)
print("Random Forest R2:", r2_rf)
```

employee salary: Linear Regression, Decision Tree, and Random Forest. The dataset was
